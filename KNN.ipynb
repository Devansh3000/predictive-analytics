### K-Nearest Neighbors (KNN) from Scratch

This simple implementation covers the basic steps of KNN:
1.  **Dataset**: A small, illustrative dataset.
2.  **Distance Function**: A way to measure how 'similar' two data points are (e.g., Euclidean distance).
3.  **Find Neighbors**: Identify the `k` closest data points to a new, unseen data point.
4.  **Prediction**: Determine the class of the new data point based on the majority class of its `k` neighbors.

```python
import numpy as np
from collections import Counter

# 1. Sample Dataset
# Each row: [feature1, feature2, ..., class_label]
dataset = np.array([
    [1, 1, 'A'],
    [1, 2, 'A'],
    [2, 2, 'A'],
    [3, 3, 'B'],
    [3, 4, 'B'],
    [4, 4, 'B']
])

X_train = dataset[:, :-1].astype(float) # Features
y_train = dataset[:, -1] # Class labels

print("Training Features (X_train):\n", X_train)
print("Training Labels (y_train):\n", y_train)
```

```python
# 2. Distance Function (Euclidean Distance)
def euclidean_distance(point1, point2):
    return np.sqrt(np.sum((point1 - point2)**2))

# Test the distance function
point_a = np.array([1, 1])
point_b = np.array([3, 4])
print(f"Distance between {point_a} and {point_b}: {euclidean_distance(point_a, point_b)}")
```

```python
# 3. Find K Nearest Neighbors
def find_k_nearest_neighbors(X_train, y_train, new_point, k):
    distances = []
    for i, train_point in enumerate(X_train):
        dist = euclidean_distance(new_point, train_point)
        distances.append((dist, y_train[i]))

    # Sort by distance and get the top k
    distances.sort(key=lambda x: x[0])
    k_nearest = distances[:k]
    return k_nearest

# 4. Make Prediction
def predict_knn(X_train, y_train, new_point, k=3):
    k_nearest = find_k_nearest_neighbors(X_train, y_train, new_point, k)

    # Get the labels of the k nearest neighbors
    neighbor_labels = [label for dist, label in k_nearest]

    # Count occurrences of each label and return the most common one
    most_common = Counter(neighbor_labels).most_common(1)
    return most_common[0][0]

# Example Usage:
# Let's try to classify a new point [2.5, 2.5]
new_point = np.array([2.5, 2.5])
k_value = 3 # Number of neighbors to consider

print(f"New point to classify: {new_point}")

# Find its neighbors
neighbors = find_k_nearest_neighbors(X_train, y_train, new_point, k_value)
print(f"\n{k_value} nearest neighbors and their labels: {neighbors}")

# Predict the class
prediction = predict_knn(X_train, y_train, new_point, k=k_value)
print(f"\nPredicted class for {new_point} (with k={k_value}): {prediction}")

# Another example
new_point_2 = np.array([1.5, 1.5])
prediction_2 = predict_knn(X_train, y_train, new_point_2, k=k_value)
print(f"Predicted class for {new_point_2} (with k={k_value}): {prediction_2}")
```
